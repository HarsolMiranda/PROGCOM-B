```python

class Paso1:
    def __init__(self):
        self.mov1 = "Mover la mano hacia arriba "
        self.mov2 = "hacer gesto de tomar soda"

    def ejecutar(self, a, b):
        for _ in range(b):
            for _ in range(a):
                print(self.mov1)
                print(self.mov2)
```

```python
p1 = Paso1()
p1.ejecutar(1, 1)
```
